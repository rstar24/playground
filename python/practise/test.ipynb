```python
import networkx as nx  
import matplotlib.pyplot as plt

gh = nx.Graph()

gh.add_edge("a","b")
gh.add_edge("b","c")
gh.add_edge("c","d")
gh.add_edge("e","d")
gh.add_edge("e","f")
gh.add_edge("a","f")

nx.draw(gh,with_labels=True)
plt.show()
```

```python
import random 
ex1 = [random.randint(1,200) * i for i in range(52)]
ex1 = list(set(ex1))
# suffle using random 
for i in range(len(ex1)):
    rd = random.randint(0,(51-i))
    temp = ex1[i]
    ex1[i] = ex1[rd]
    ex1[rd]  = temp

print(len(ex1))
print(len(set(ex1)))

# convert the list into a binary search tree

```
